### Startup Acquisition Prediction - EDA

#### Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("data/startup_success_raw.csv")

df.head(5)
```

```python
df.dtypes
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.plot(kind="box",subplots=True,layout=(1,11),figsize=(20,3))
plt.tight_layout()
```

```python
outlier_columns = ['funding_rounds','market_size_billion','product_traction_users','burn_rate_million','revenue_million']

for col in outlier_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    minimum = q1 - (1.5 * iqr)
    maximum = q3 + (1.5 * iqr)
    df = df[(df[col] >= minimum) & (df[col] <= maximum)]

df
```

```python
df.plot(kind="box",subplots=True,layout=(1,11),figsize=(20,3))
plt.tight_layout()
```
